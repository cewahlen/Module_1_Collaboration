There are 3600 seconds in one hour because there are 60 seconds in one minute and 60 minutes in one hour

```python
seconds_per_hour = 60 * 60
```

The variable seconds_per_hour stores the number of seconds in one hour, which is 3600

```python
seconds_per_hour * 24
```

There are 86,400 seconds in one day

```python
seconds_per_day = seconds_per_hour * 24
```

The variable seconds_per_day stores the number of seconds in one day, which is 86,400

```python
seconds_per_day / seconds_per_hour
```

Using floating-point division, the result is 24.0, meaning there are 24 hours in one day.

```python
seconds_per_day // seconds_per_hour
```

Yes. The integer division result (24) agrees with the floating-point division result (24.0). The only difference is that floating-point division includes the decimal point and zero.
